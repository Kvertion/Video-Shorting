```python
def reading(path_to,chapters=1,duration=10):
    # duration=28215.2
    # chapters=1
    # if chapters==1:
    #     time_convert=[0.0]
    # else:
    time_convert=[round((duration*i)/chapters,2) for i in range(0,chapters)]
    dfs=[pd.read_csv(f'{path_to}_{i}.csv') for i in range(1,chapters+1)]
    # print([f'{i}' for i in range(1,chapters+1)])
    for i in range(chapters):
        dfs[i]["0"]+=time_convert[i]
        dfs[i]["1"]+=time_convert[i]
    df=pd.concat(dfs,ignore_index=True)
    # df=pd.read_csv('ИГРОПОЛИУС2025_5_.csv')
    df.columns=['id','start','end','text']
    # df=df.set_index('id')
    df.head()
    return df,dfs

def cleaning(df):
    #old
    # df=df[(df['text'].str.contains('DimaTorzok')==False) & (df['text'].str.contains('Продолжение следует')==False)]
    # df['text_lower']=df['text'].apply(lambda x: x.lower().replace(' ','').replace(',','').replace('.','').replace('?','').replace('!','').replace(':',''))
    # # df['is_in_text']=df['text_lower'].apply(lambda x: x in phrase)
    # df[df['text_lower']!=''].reset_index(drop=True,inplace=True)
    # # df['is_in_text'].describe()
    
    #new

    df=df[(df['text'].str.contains('DimaTorzok')==False) & (df['text'].str.contains('Продолжение следует')==False)]
    df['text']=df['text'].str.replace(r'[^a-яА-ЯёЁa-zA-z ]','',regex=True)
    df['text_lower']=df['text'].apply(lambda x: x.lower())
    df[df['text']!=''].reset_index(drop=True,inplace=True)
    df[df['text_lower']!=''].reset_index(drop=True,inplace=True)

    return df

def all_text(df):
    return df['text'].sum()

def get_timecode(df,phrases):
    all_timecode=[]
    for phrase in phrases:
        df['is_in_text']=df['text_lower'].apply(lambda x: x in phrase)
        df_text=df[df['is_in_text']]
        indexes=dict(enumerate(df_text.index))
        good_indexes=[]
        phrase_finder={}
        for i in range(len(indexes)-1):
            if df.loc[indexes[i],'text_lower']+df.loc[indexes[i+1],'text_lower'] in phrase:
                j=i
                dec=df.loc[indexes[j],'text_lower']
                try:
                    while dec in phrase:
                        j+=1
                        dec+=df.loc[indexes[j],'text_lower']
                    # print(dec)
                    if j not in phrase_finder.keys():
                        phrase_finder[j]=[indexes[i]]
                    else:
                        phrase_finder[j]+=[indexes[i]]
                except KeyError:
                    print(f'{KeyError}')
        my_phrase=[]
        # print(phrase_finder)
        for i in phrase_finder.values():
            # print(i)
            if len(i)>len(my_phrase):
                my_phrase=i
        
        if len(my_phrase)>0:
            my_phrase=[my_phrase[0]-1]+my_phrase+[my_phrase[-1]+1]
            # my_phrase.append(my_phrase[-1]+1)
            # print(my_phrase)
            end_point=[df.loc[j,'text_lower'] for j in my_phrase]
            timecode=[df.loc[my_phrase[0],'start'],df.loc[my_phrase[-1],'end']]
            all_timecode.append(timecode)
    return all_timecode
```
