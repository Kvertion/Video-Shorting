```python
from google.colab import drive
import pandas as pd
drive.mount('/content/drive')
```

```python
from faster_whisper import WhisperModel
model = WhisperModel("large-v3", device="cuda", compute_type="float16")
names=[f'audio_{i}.wav' for i in range(1,9)]
for name in names:
  segments, info = model.transcribe(f"/content/drive/My Drive/{name}", language='ru', vad_filter=True)
  # print("".join([segment.text for segment in segments]))
  result_list=[]
  for i in segments:
    x=[i.start,i.end,i.text]
    print(x)
    result_list.append(x)
  df=pd.DataFrame(result_list)
  df.to_csv(f"/content/drive/MyDrive/{name.replace('wav','csv')}")
  print('completed')
```
